# Optimizing the Performance of Gold Tables

The purpose of this notebook is to optimize the performance of gold-layer tables.

The following operations were performed:
1. Ran `OPTIMIZE` with `ZORDER BY` to reduce small-file overhead, and collocate related records in each table.
2. Displayed the version history of `customer_product_matrix_gold`.
3. Time-traveled `customer_product_matrix_gold` table to version 2.
4. Enabled Change Data Feed on all tables.

### Optimizing customer product matrix gold table

```python
%sql
OPTIMIZE workspace.retail_schema.customer_product_matrix_gold ZORDER BY (Purchase_Count)
```

### Optimizing customer segmentation by city gold table

```python
%sql
OPTIMIZE workspace.retail_schema.customer_segmentation_city_gold ZORDER BY (Total_Customers)
```

### Optimizing customer segmentation by region gold table

```python
%sql
OPTIMIZE workspace.retail_schema.customer_segmentation_region_gold ZORDER BY (Region)
```

### Optimizing product sales gold table

```python
%sql
OPTIMIZE workspace.retail_schema.product_sales_gold ZORDER BY (Total_No_Of_Orders)
```

### Optimizing sales by customer gold table

```python
%sql
OPTIMIZE workspace.retail_schema.sales_by_customer_gold ZORDER BY (Region)
```
